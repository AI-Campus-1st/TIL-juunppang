### 문항 1 CSV 다루기

```python
SUBJECTS = ["kor", "eng", "math"]

def to_score(name, subject, raw):
    try:
        return int(raw)
    except ValueError:
        print(f"[처리] {name} - {subject} 값 오류({raw}) → 0점 처리")
        return 0

students = []
with open('scores.csv', 'r') as f:
	columns = f.readline().split(',')

	for row in f.readlines():
		temp = row.split(',')
		name = temp.pop(0)
		scores = [to_score(name, s, temp[i]) for i, s in enumerate(SUBJECTS)]

		students.append({
			'name': name,
			'avg': round(sum(scores)/len(scores), 1)
		})

with open('result.csv', 'w') as f:
	f.write('name,avg\n')
	for student in students:
		name = student['name']
		avg = student['avg']
		f.write(f'{name},{avg}\n')

print("완료: result.csv 생성")
```

### 문항 2 CSV를 JSON으로 변환

```python
import json

members = []
errors = []

with open('members.csv', 'r') as f:
    
    columns = f.readline().split(',')
    for i, row in enumerate(f.readlines(), start=2):
        mid, name, age = row.split(',')
        try:
            if not name:
                raise ValueError('name 비어있음')
            member = dict(zip(columns, [int(mid), name, int(age)]))
            members.append(member)
        except ValueError as e:
            errors.append(f'{i}행: {e}')

with open('members.json', 'w') as f:
    json.dump(members, f, ensure_ascii= False, indent=2)

if errors:
    with open('errors.log', 'w') as f:
        f.write("\n".join(errors))

print(f'성공 {len(members)}건 / 실패 {len(errors)}건')
```
