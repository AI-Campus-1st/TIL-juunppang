![3번 문제](<attachment:Screenshot 2026-07-27 215359.png>)

```python
num = int(input())
ans = 0 
for i in range(num):
    nums = i + 1
    ans += nums
print(ans)
```

![4번 문제](<attachment:Screenshot 2026-07-27 220048.png>)

```python
int_n = int(input())
for i in range(1, int_n+1):
    if int_n % i == 0:
        nums = i
        print(nums, end=' ')
```

![6번 문제](<attachment:Screenshot 2026-07-27 221517.png>)

```python
num = input()
nums = range(1,int(num)+1)
for i in nums:
    for j in range(1,i+1):
        print(j, end = ' ')
    print()
                
```
