### 문항 1 Numpy 배열로 5명 학생의 4과목 성적표를 만들고 축(axis) 기준 통계를 산출하시오.

아래 1차원 리스트는 학생 5명이 4개 과목(국어, 영어, 수학, 과학 순)에서 받은 점수를 이어 붙인 것입니다.

Numpy만 사용하여(반복문 사용 금지) 다음을 순서대로 수행하고 결과를 출력하세요.
```py
scores = [88, 92, 79, 95,
          75, 68, 82, 70,
          92, 99, 95, 88,
          60, 72, 65, 58,
          85, 80, 78, 90]
```

````python
# 1 위 리스트를 5행 4열 ndarray로 변환하고 shape, ndim, size, dtype을 출력하시오.
import numpy as np

scores = [88, 92, 79, 95,
          75, 68, 82, 70,
          92, 99, 95, 88,
          60, 72, 65, 58,
          85, 80, 78, 90]

s = np.array(scores).reshape(5, 4)
print(s)
print(s.shape, s.ndim, s.size, s.dtype)
````

````python
# 2 학생 별 평균과 과목 별 평균을 각각 구하시오.
student_avg = np.mean(s, axis=1)
subjects_avg = np.mean(s, axis=0)
print(student_avg) # 학생 별 평균 점수
print(subjects_avg) # 교과 별 평균 점수
````

````python
# 3 전체에서 가장 높은 점수와 그 점수가 몇 번째 학생·몇 번째 과목인지 구하시오.
max_scores = np.max(s) # 전체 Max 값
max_arg = np.argmax(s) # Max 값 위치
max_pos = np.argwhere(s == max_scores)
print(max_scores)
print(max_pos)
````

````python
# 4 전체 평균, 표준편차, 분산을 구하시오.
x = s.mean() # 전체 평균
y = s.std() # 전체 표준편차
z = s.var() # 전체 분산
print(x, y, z)
````

````python
# 5 평균이 80점 이상인 학생의 인덱스와 인원수를 구하고, 해당 학생들의 성적만 잘라내어 출력하시오.
s_idx = np.where(student_avg >= 80) # 평균 80점 이상 학생 인덱스
print(s_idx, np.sum(student_avg >= 80)) # 인덱스와 인원수
print(s[s_idx]) # 평균 80점 이상 학생 점수만 출력
````

````python
# 6 90점 이상인 점수의 개수, 100점 이상인 점수가 하나라도 있는지, 모든 점수가 60점 이상인지를 판별하시오.
print(np.sum( s >= 90 ), np.sum( s >= 100 ), np.all( s >= 60))
````

````python
# 7 90점 이상은 'A', 80점 이상은 'B', 나머지는 'C'로 변환한 등급 배열을 만드시오.
grade = np.where(s>= 90, 'A', np.where(s >=80, 'B', 'C'))

print(grade)
````

### 문항 2 8x8 난수 배열을 흑백 이미지로 가정하고 슬라이싱 이진화 좌우 반전을 구현하시오.

0~255 사이의 정수 난수로 이루어진 8행 8열 배열을 흑백 이미지 데이터라고 가정합니다. 채점 결과를 동일하게 맞추기 위해 반드시 아래 시드를 먼저 실행하세요.
```py
import numpy as np
np.random.seed(2024)
img = np.random.randint(0, 256, size=(8, 8))
```

````python
import numpy as np
np.random.seed(2024)
img = np.random.randint(0, 256, size=(8, 8))
````

````python
# 1 생성한 img를 출력하시오.
print(img)
````

````python
# 2 이미지 정중앙의 4×4 영역만 잘라내어 출력하시오.
print(img[2:6, 2:6])
````

````python
# 3 짝수 번째 행(0, 2, 4, 6행)만 뽑아내시오.
print(img[::2])
````

````python
# 4 128 이상인 픽셀은 255(흰색), 미만인 픽셀은 0(검은색)으로 바꾼 이진화 배열을 만드시오.
bw = np.where(img >= 128, 255, 0)
print(bw)
````

````python
# 5 이진화 결과에서 흰색 픽셀과 검은색 픽셀의 개수를 각각 세시오.
print(np.sum(bw == 255), np.sum(bw == 0))
````

````python
# 6 원본 이미지를 좌·우 절반으로 분할하시오.
img_split =np.split(img,2,axis=1)
print(f'원본 이미지 좌: \n {img_split[0]}')
print(f'원본 이미지 우: \n {img_split[1]}')
````

````python
# 7 분할한 두 조각의 순서를 바꿔 다시 이어 붙여 좌우가 뒤바뀐 이미지를 만드시오.
rev_img = np.concatenate([img_split[1], img_split[0]], axis=1)
print(rev_img)
````

````python
# 8 원본 이미지를 1차원으로 펼친 뒤 앞 10개 값과 shape을 출력하시오.
one_dim_img =img.ravel()
print(one_dim_img[:10], one_dim_img.shape)
````

````python
# 9 128 이상인 픽셀만 골라내어 그 개수와 평균값을 구하시오.
cnt = np.sum(img >= 128)
print(cnt)
img[img < 128 ] = 0
avg = np.sum(img) / cnt
print(avg)
````

````python
cnt1 = img[img >= 128]
cnt1
print(np.size(cnt1), np.mean(cnt1))
````

### 문항 3 지점별 판매 데이터에 브로드캐스트와 행렬 연산을 적용해 매출을 분석하시오.
```py
sales = np.array([[120,  80,  45,  60],     # A지점
                  [200, 150,  90,  30],     # B지점
                  [ 75,  60, 120,  95]])    # C지점
price = np.array([1500, 3000, 5000, 2000])  # 제품 1~4 단가
```

````python
sales = np.array([[120,  80,  45,  60],     # A지점
                  [200, 150,  90,  30],     # B지점
                  [ 75,  60, 120,  95]])    # C지점
price = np.array([1500, 3000, 5000, 2000])  # 제품 1~4 단가
````

````python
#1 브로드캐스트를 이용해 각 지점의 제품별 매출액(수량 × 단가) 행렬을 구하시오. (반복문 금지)
print(sales * price)
````

````python
#2 행렬곱을 이용해 각 지점의 총매출을 한 번에 구하시오. 또한 제품별 총판매량도 구하시오.
total_sales = sales.dot(price)
print(total_sales)
per_sales = sales.sum(axis=0)
print(per_sales)
````

````python
#3 각 제품의 평균 판매량을 구한 뒤, 브로드캐스트로 평균 대비 편차 행렬을 구하시오.
avg_sales = np.mean(sales, axis=0)
print(avg_sales)
print(sales - avg_sales)
````

````python
#4 각 지점을 기준으로 판매량을 정규화하시오. 결과의 행별 평균이 0, 표준편차가 1이 되는지 확인하시오.
per_mean = np.mean(sales, axis=1, keepdims=True)
print(per_mean)
per_std = np.std(sales, axis=1, keepdims=True)
print(per_std)
sales_normalized = (sales - per_mean) / per_std
print(sales_normalized)
````

````python
print('행별 평균:', np.mean(sales_normalized, axis=1).round(2))
print('행별 표준편차:', np.std(sales_normalized, axis=1).round(2))
````

````python
#5 sales의 전치행렬을 구하고 shape을 출력하시오.
print(sales.T) # 전치행렬
print(sales.T.shape) # 전치행렬의 shape
````

````python
#6 sales의 앞 3개 열만 잘라 3×3 정방행렬 A를 만들고, 행렬식과 역행렬을 구하시오. 이어서 A와 역행렬을 곱해 단위행렬이 나오는지 확인하시오.
A = sales[:3, :3] #정방행렬
print(A)
````

````python
x = np.linalg.det(A)
print(x) #행렬식
````

````python
inv_A = np.linalg.inv(A)
print(inv_A) # A의 역행렬
````

````python
print((A @ inv_A).round(1)) # 단위행렬
````

````python
#7 총매출이 가장 높은 지점의 인덱스와, 지점 순서대로의 누적 매출을 구하시오.
print('총매출이 가장 높은 지점의 인덱스:', np.argmax(total_sales))
print('지점 순서대로 누적 매출:', np.cumsum(total_sales))
````
