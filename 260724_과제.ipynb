문항 1) 복사가 잘 된 건가?

아래의 코드를 실행 하였을 때, 변수 a와 b에 담긴 list의 요소가 같은지 혹은 다른지 여부를 판단하고 그 이유를 작성하시오.

```python
a = [1, 2, 3, 4, 5]
b = a

a[2] = 5

print(a)
print(b)
```

위의 a,b의 출력 값이 같은 이유는, list를 .cpoy() 혹은 .deepcopy(새로운 주소로 list를 복사해주는 메서드)로 복사하지 않으면, 결국 같은 주소를 바라보게 되는 list로 복사하기 때문에 결과 값이 같을 수 밖에 없다.


문항 2) 평균 점수 구하기

key 값으로 과목명, value 값으로 점수를 가지는 scores의 전체 과목의 평균 점수를 계산하시오.

scores = {'python': 80, 'algorithm' : 90, 'django' : 89, 'web' : 83} # => 85.5

```python
scores = {'python': 80, 'algorithm' : 90, 'django' : 89, 'web' : 83}
key_cnt = 0
for key in scores.keys():
    key_cnt += 1
    print(key_cnt)

sum(scores.values()) / key_cnt
```

문항 3)

여러 사람의 혈액형(A, B, AB, O)에 대한 정보가 담긴 bloods를 전달 받아, key는 혈액형, value는 사람 수인 dictionary를 만드시오.

```python
bloods = [
    'A', 'B', 'A', 'O', 'AB', 'AB',
    'O', 'A', 'B', 'O', 'B', 'AB',
] # => {'A': 3, 'B': 3, 'O':  3, 'AB': 3}
```

```python
blood_info = {}
for key in bloods:
    value = bloods.count(key)
    blood_info[key] = value
print(blood_info)
    
```
