문항 1) 모음은 몇 개나 있을까?

문자열이 주어졌을 때, 해당 문자열의 모음 갯수를 계산하는 코드를 작성하시오.

    word1 = 'apple' # => 2
    word2 = 'banana' # => 3


```python
word = input()
print(word)
vowels = ['a', 'e', 'i', 'o', 'u']
vowel_cnt = 0
for i in vowels:
    cnt = word.count(i)
    vowel_cnt += cnt
print(vowel_cnt)
```

문항 2) 네모 출력
두 개의 정수 n과 m이 주어졌을 때, 가로의 길이가 n, 세로의 길이가 m인 직사각형 형태를 별(*) 문자를 이용하여 출력하시오.

단, 반복문은 사용할 수 없다.

    n = 5 
    m = 9

```python
n = 5 
m = 9
rectangle = ((('*'*n)+'\n')*m)
print(rectangle)
```

문항 3) Built-in 함수와 메서드

sorted()와 .sort()의 차이점을 코드의 실행 결과를 활용하여 설명하시오.

```python
import random
nums = list(range(0,100))
lotto = random.sample(nums, 6)
#sorted_lotto = sorted(lotto)

sorted_lotto = lotto.sort()
print(lotto)
print(sorted_lotto)
```

```python
import random
nums = list(range(0,100))
lotto = random.sample(nums, 6)
sorted_lotto = sorted(lotto)

#sorted_lotto = lotto.sort()
print(lotto)
print(sorted_lotto)
```

차이점은 내장함수 sorted()는 원본 list을 유지하고, 정렬한 결과를 가지고, 메서드 .sort()는 원본 list를 변형시키고, None을 리턴합니다.
