문항 2)  반복문으로 네모 출력

두 개의 정수 n과 m이 주어졌을 때, 가로의 길이가 n, 세로의 길이가 m인 직사각형 형태를 별(*) 문자를 이용하여 출력하시오. 단, 반복문을 사용하여 작성하시오.

n = 5   
m = 9 

```python
n = input()
m = input()

for i in range(int(m)):
    for j in range(int(n)):
        print('*', end = '')
    print('')
```
