### 문항 1 재귀 함수

재귀 함수를 사용했을때 얻을 수 있는 장점과 단점을 반복문과 비교하여 작성하시오.


````python
# 피보나치 수열 재귀 함수 구현  
def fib(n):
    if n in [1,2]:
        return 1
    return fib(n-1) + fib(n-2)


# 피보나치 수열 반복문으로 구현
def fib_loop(n):
    if n == 1 or n == 2:
        return 1
    a = b = 1
    for i in range(n-2):
        b, a = a+b, b
    return b
````

````python
fib(10)
````

````python
fib_loop(10)
````

> 장점 코드가 더 짧고, 간결하다. 

> 단점 실제로 fib(2)의 함수가 34개, fib(1)의 함수가 21개 총 55개의 함수 stack이 쌓여서 속도가 느리고 메모리 사용량이 많다.

> 반대로 반복문은 하나의 함수 안에서 반복이 이뤄져 결과를 도출한다. 함수 안이 복잡해질 수 있으나, 함수의 갯수가 적어 메모리 사용량이 적고 빠르다.

### 문항 2 Circle 클래스 생성, 활용

다음의 요구사항을 만족하는 Circle 클래스를 정의하고, 넓이와 둘레, 중심 좌표를 구하는 메서드를 정의하시오.

```py
# 반지름 : 3, x좌표: 2, y좌표: 4
c1 = Circle(3, 2, 4)

c1.area() #=> 28.26
c1.circumference() #=> 18.84
c1.center() #=> (2, 4)

````python
class Circle:
    def __init__(self, r, x, y):
        self.r = r
        self.x = x
        self.y = y
    def area(self):
        return 3.14*self.r**2
    def circumference(self):
        return 2*3.14*self.r
    def center(self):
        return self.x, self.y
````

````python
c1 = Circle(3, 2, 4)
````

````python
c1.area()
````

````python
c1.circumference()
````

````python
c1.center()
````
