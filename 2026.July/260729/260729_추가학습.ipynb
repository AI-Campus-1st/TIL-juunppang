### 1. 숫자의 의미

정수로 이루어진 list를 전달 받아, 각 정수에 대응되는 아스키 문자를 이어붙인 문자열을 반환하는 get_secret_word 함수를 작성하시오. 단, list는 65이상 90이하 그리고 97이상 122이하의 정수로만 구성되어 있다.

```py
get_secret_word([83, 111, 115, 108, 110]) #=>SosIn
```

````python
def get_secret_word(nums):
    char = []
    for num in nums:
        char.append(chr(num))
    return ''.join(char)

````

````python
get_secret_word([83, 111, 115, 108, 110])
````

### 2. 내 이름은 몇일까?

문자열을 전달 받아 해당 문자열의 각 문자에 대응되는 아스키 숫자들의 합을 반환하는 get_secret_number 함수를 작성하시오. 단, 문자열은 A~Z, a~z로만 구성되어 있다.

```py
get_secret_number('tom') #=> 336


````python
def get_secret_number(name):
    sum = 0
    for i in name:
        temp = ord(i)
        sum += temp
    return sum 
````

````python
get_secret_number('tom')
````

### 3. 강한 이름

문자열 2개를 전달 받아 두 문자열의 각 문자에 대응되는 아스키 숫자들의 합을 비교하여 더 큰 합을 가진 문자열을 반환하는 get_strong_word 함수를 작성하시오.

```py 
get_strong_word('z', 'a') #=> 'z'
get_strong_word('tom', 'john') #=> 'john

````python
def get_strong_word(*char):
    total_sum = 0
    tmp_list1 = []
    tmp_list2 = []
    for i in range(2): 
        total_sum = 0
        tmp_list1 = []
        tmp_list1.extend(list(char[i]))

        for j in range(len(tmp_list1)):
            total_sum += ord(tmp_list1[j])
        tmp_list2.append(total_sum)

    if tmp_list2[0] > tmp_list2[1]:
        return char[0]
    else: 
        return char[1]
````
